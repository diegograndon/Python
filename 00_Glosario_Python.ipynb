## 1. Conceptos y comandos

### `print()`

**¿Qué hace?**  
Muestra información en pantalla.

**Sintaxis:**

```python
print(valor)
```

**Ejemplo:**

````python
nombre = "Ana"
print(nombre)
````

### `input()`

**¿Qué hace?**  
Permite recibir información escrita por el usuario.

**Importante:** `input()` devuelve un `str` (texto), aunque el usuario escriba un número.

**Ejemplo:**

````python
edad = input("¿Qué edad tienes? ")
print(edad)
````

### `int()`

**¿Qué hace?**  
Convierte un valor a un número entero cuando la conversión es posible.

**Sintaxis:**

```python
int(valor)
```

**Ejemplo:**

````python
edad = int(input("¿Qué edad tienes? "))
print(edad + 1)
````

## 2. Sintaxis importante

### Condicional `if`

**¿Para qué sirve?**  
Permite ejecutar instrucciones cuando una condición es verdadera.

**Estructura:**

```python
if condicion:
    instrucciones
```

**Ejemplo:**

````python
edad = 17

if edad >= 18:
    print("Mayor de edad")
else:
    print("Menor de edad")
````

### `=` versus `==`

**`=`** significa **asignación**: guarda un valor en una variable.

```python
x = 5
```

**`==`** significa **comparación**: pregunta si dos valores son iguales.

```python
x == 5
```

**Idea clave:** en Python, `=` y `==` cumplen funciones diferentes.

## 3. Errores que debemos reconocer

### `NameError`

**¿Qué significa?**  
Python no encuentra el nombre que estamos intentando utilizar.

**Ejemplo:**

````python
print(nombre_usuario)
````

**¿Qué debo revisar?**

- ¿Escribí correctamente el nombre de la variable?
- ¿La variable fue creada antes de utilizarla?

### `TypeError`

**¿Qué significa?**  
Intentamos realizar una operación que no es válida para los tipos de datos involucrados.

**Ejemplo:**

````python
edad = "15"
print(edad + 1)
````

**Una posible solución:** convertir el texto a entero.

```python
edad = int(edad)
print(edad + 1)
```
